```python
# Exercise 4: nested for loops
for row in range(8):
    for column in range(8):
        print('#', end=' ')
    print()
```

```python
row = 0
while row < 8:
    column = 0
    while column < 8:
        print('#', end=' ')
        column += 1
    print()
    row += 1
```

```python
# Exercise 9: sum all numbers from 0 to 100
total = 0
for number in range(101):
    total += number
print('The sum of all numbers is', total)
```

```python
# Exercise 10: sum even and odd numbers separately
even_sum = 0
odd_sum = 0
for number in range(101):
    if number % 2 == 0:
        even_sum += number
    else:
        odd_sum += number
print(f'Sum of all evens: {even_sum}')
print(f'Sum of all odds: {odd_sum}')
```

```python
# Exercise 13: load country data and count languages
import json
from collections import Counter
from pathlib import Path

project_folder = Path.cwd().resolve()
while not (project_folder / 'data').is_dir() and project_folder != project_folder.parent:
    project_folder = project_folder.parent

if not (project_folder / 'data').is_dir():
    raise FileNotFoundError(f'Could not find the data folder from {Path.cwd()}')

data_file = project_folder / 'data' / 'countries-data.py'

with open(data_file, 'r', encoding='utf-8') as file:
    countries_data = json.load(file)

total_languages = sum(len(country.get('languages', [])) for country in countries_data)
print('Total number of languages:', total_languages)

language_counter = Counter()
for country in countries_data:
    for language in country.get('languages', []):
        language_counter[language] += 1

print('Ten most spoken languages:')
for language, count in language_counter.most_common(10):
    print(f'{language}: {count}')
```

### Learning: JSON data, dictionaries, and `dict.get()`

The `countries-data.py` file contains JSON data: a list of dictionaries. `json.load(file)` converts that file into normal Python objects.

`country.get("languages", [])` safely reads the `languages` key. If the key is missing, it returns an empty list instead of raising a `KeyError`.

```python
first_country = countries_data[0]

print(first_country)
print(first_country.get("languages", []))
print("Number of languages:", len(first_country.get("languages", [])))
print("Country type:", type(first_country))
print("All data type:", type(countries_data))
```

### Learning: `Counter` and nested loops

`Counter` stores frequency counts like a dictionary. The outer loop visits each country, and the inner loop visits each language in that country. Each language count is increased by one. `most_common(10)` returns the ten languages used by the greatest number of countries.

```python
# Exercise 14: ten most populated countries
population_counter = Counter()
for country in countries_data:
    population_counter[country.get('name', 'Unknown')] += country.get('population', 0)

print('Ten most populated countries:')
for country, population in population_counter.most_common(10):
    print(f'{country}: {population}')
```
